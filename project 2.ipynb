```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Anu", "Meera", "Asha", "Diya", "Riya",
             "Neha", "Sara", "Ammu", "Niya", "Maya"],
    "Age": [22, 24, 23, 21, 25, 22, 24, 23, 21, 25],
    "Maths": [85, 90, 95, 45, np.nan, 88, 76, 80, 92, 70],
    "Science": [80, 85, 90, 55, 65, np.nan, 78, 82, 88, 75],
    "City": ["Kannur", "Kochi", "Kannur", "Calicut", "Kochi",
             "Kannur", "Calicut", "Kochi", "Kannur", "Calicut"]
}

df = pd.DataFrame(data)
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print(df)

print(df.columns)
print(df.head(10))
print(df.duplicated().sum())

df = df.drop_duplicates()

print(df.isnull().sum())
print(df.dropna())

df_mean = df.copy()
df_mean[["Maths", "Science"]] = df_mean[
    ["Maths", "Science"]
].fillna(df_mean[["Maths", "Science"]].mean())
print(df_mean)

df_median = df.copy()
df_median[["Maths", "Science"]] = df_median[
    ["Maths", "Science"]
].fillna(df_median[["Maths", "Science"]].median())
print(df_median)

df = df_mean.copy()

print(df[["Name", "Maths"]])
print(df.loc[df["Age"] > 22])
print(df.iloc[0:5])

print(np.mean(df["Maths"].to_numpy()))
print(np.sqrt(df["Maths"].to_numpy()))

def add_bonus(dataframe):
    result = dataframe.copy()
    result["Bonus"] = result["Maths"] + 5
    return result

df = df.pipe(add_bonus)
print(df)

print(np.union1d(df["Maths"].to_numpy(), df["Science"].to_numpy()))
print(np.intersect1d(df["Maths"].to_numpy(), df["Science"].to_numpy()))
print(np.setdiff1d(df["Maths"].to_numpy(), df["Science"].to_numpy()))

print(df.sort_values(by="Maths", ascending=True))
print(df.sort_values(by="Maths", ascending=False))

print(df["Maths"].max())
print(df["Maths"].min())
print(df["Maths"].std())

print(df[df["Maths"] > 80])

df["Total"] = df["Maths"] + df["Science"]
print(df)

print(df["City"].unique())
print(df["City"].nunique())
```
